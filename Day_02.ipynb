### For loop

```python
for i in range(10,0,-1):
    print(i,end=" ")
```

```python
for i in range(9):
    print("Hello")
```

```python
while True:
    print("Hello")
```

```python
n=10
while n!=0:
    print(n,end=" ")
    n-=1
else:
    print("\nValue should not be zero")
```
