Boolean Masking

Boolean masking is a NumPy technique that uses a Boolean array (an array of True and False values) to select, filter, or modify elements in another array based on a specified condition.

```python
arr=np.array([80,10,30,40,65])
bool_mask=arr>50
arr[bool_mask]
```

```python
arr=np.array([80,10,30,40,65])
bool_mask=arr>20
arr[bool_mask]
```

```python
order_amount=1000 1500 800 3000 4000 900
1.filter order_amount>2000
2.order_amount>2000 must be multiply by 2 
3.add new order_amount at the end 5000
4.print the last order amount
```

```python
import numpy as np

# Create the array
order_amount = np.array([1000, 1500, 800, 3000, 4000, 900])

# 1. Filter order_amount > 2000
print("1. Order amounts > 2000:")
print(order_amount[order_amount > 2000])

# 2. Multiply order_amount > 2000 by 2
order_amount[order_amount > 2000] *= 2
print("\n2. After multiplying by 2:")
print(order_amount)

# 3. Add new order_amount at the end (5000)
order_amount = np.append(order_amount, 5000)
print("\n3. After adding 5000:")
print(order_amount)

# 4. Print the last order amount
print("\n4. Last order amount:")
print(order_amount[-1])
```

```python
order_amount=np.array(list(map(int,input("Enter the amount:").split())))
print(order_amount)
print("order>2000:",order_amount[order_amount>2000])
print("order>2000 multi of 2:")
print(np.where(order_amount>2000,order_amount*2,order_amount))
order_amount=np.append(order_amount,5000)
print("new order added:",order_amount)
print("last order amount:",order_amount[-1])
```
