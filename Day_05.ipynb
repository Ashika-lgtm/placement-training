```python
'''File-used for data storing
File handling is a concept where we can create,read and write a file.
File operation modes:
'w'-write
'r'-read
'a'-append
'w+'-write and read
'r+'-read and write
'''
```

```python
with open("sample.txt","w") as file:
    file.write("\nHey there!")
```

```python
with open("sample.txt","a") as file:
    file.write("\nHey! this is sample txt")
```

```python
with open("sample.txt","a") as file:
    file.write("\nHey there!")
```

```python
with open("sample.txt","w") as file:
    file.writelines('''
    okay everything is cool here
    file handling is good to use
    when we have less data
    ''')
```

```python
with open("sample.txt","r") as file:
    data=file.read()
    print(data)
```

```python
with open("demo1.txt","w+") as file:
    file.writelines('''    
    Hey there!,
    This is sample text file!,
    Hurray!!!!
    ''')
    file.seek(0)      #it points to the current cursor
    data=file.read()
print(data)
```

### FILE READING

''''''''''''
file operation mode
'w'-write
'r'-read
'a'-append
'w+'-write and read
'r+'-read and write
'x'-open file in exclusive mode(throw error if file already exists)
'rb'-reading the byte data
'wb'-writing byte data

1.write()
2.writelines()
3.read()
4.readlines()
5.seek()-pointer movement
6.tell()-pointer position
''''''''

```python
file=open("sample.txt",'r')
data=file.read()
file.close()
print(data)
```

```python
with open("sample.txt","r")as file:
    file.seek(20)#
    print(file.tell())#get the pointer location
```
