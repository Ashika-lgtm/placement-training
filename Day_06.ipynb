```python
import customtkinter as ctk

# -----------------------
# Configuration
# -----------------------
ctk.set_appearance_mode("dark")     # dark / light / system
ctk.set_default_color_theme("blue") # blue / green / dark-blue

# -----------------------
# Window
# -----------------------
app = ctk.CTk()
app.title("Login")
app.geometry("500x500")
app.resizable(False, False)

# -----------------------
# Login Function
# -----------------------
def login():
    username = username_entry.get()
    password = password_entry.get()

    if username == "admin" and password == "1234":
        status_label.configure(text="Login Successful!", text_color="green")
    else:
        status_label.configure(text="Invalid Username or Password", text_color="red")

# -----------------------
# Main Frame
# -----------------------
frame = ctk.CTkFrame(app, width=380, height=420, corner_radius=20)
frame.place(relx=0.5, rely=0.5, anchor="center")

# -----------------------
# Heading
# -----------------------
title = ctk.CTkLabel(
    frame,
    text="Welcome Back",
    font=("Arial", 28, "bold")
)
title.pack(pady=(35, 5))

subtitle = ctk.CTkLabel(
    frame,
    text="Sign in to continue",
    text_color="gray"
)
subtitle.pack(pady=(0, 25))

# -----------------------
# Username
# -----------------------
username_entry = ctk.CTkEntry(
    frame,
    width=250,
    height=40,
    placeholder_text="Username"
)
username_entry.pack(pady=10)

# -----------------------
# Password
# -----------------------
password_entry = ctk.CTkEntry(
    frame,
    width=280,
    height=40,
    placeholder_text="Password",
    show="*"
)
password_entry.pack(pady=10)

# -----------------------
# Remember Me
# -----------------------
remember = ctk.CTkCheckBox(frame, text="Remember Me")
remember.pack(anchor="w", padx=45, pady=(5, 15))

# -----------------------
# Login Button
# -----------------------
login_btn = ctk.CTkButton(
    frame,
    text="Login",
    width=280,
    height=40,
    corner_radius=10,
    command=login
)
login_btn.pack(pady=10)

# -----------------------
# Status
# -----------------------
status_label = ctk.CTkLabel(frame, text="")
status_label.pack(pady=10)

# -----------------------
# Footer
# -----------------------
footer = ctk.CTkLabel(
    frame,
    text="Forgot Password?",
    text_color="#4EA8DE",
    cursor="hand2"
)
footer.pack(pady=(15, 5))
app.mainloop()
```
